**3.15. Các phép toán trên vector và ma trận**

**3.17. Môi trường tính toán Graph trong tensorflow**

```python
import tensorflow as tf
import numpy as np

def a_regular_function(x, y, b):
  x = tf.matmul(x, y)
  x = x + b
  return x

a_function_that_uses_a_graph = tf.function(a_regular_function)

x1 = tf.constant([[1.0, 2.0]])
y1 = tf.constant([[2.0], [3.0]])
b1 = tf.constant(4.0)

orig_value = a_regular_function(x1, y1, b1).numpy()
tf_function_value = a_function_that_uses_a_graph(x1, y1, b1).numpy()

print(orig_value)
print(tf_function_value)
```

```python
import tensorflow as tf
import numpy as np

def a_regular_function(x, y, b):
  x = tf.matmul(x, y)
  x = x + b
  return x

@tf.function
def outer_function(x):
  y = tf.constant([[2.0], [3.0]])
  b = tf.constant(4.0)

  return a_regular_function(x, y, b)

rs = outer_function(tf.constant([[1.0, 2.0]])).numpy()

print(rs)
```

**3.18. Modules, Layers và Models**

```python
import tensorflow as tf
import numpy as np

class SimpleModule(tf.Module):
    def __init__(self, name=None):
        super().__init__(name=name)
        self.a_variable = tf.Variable(5.0, name="train_me")
        self.non_trainable_variable = tf.Variable(5.0, trainable=False, name="do_not_train_me")

    def __call__(self, x):
        return self.a_variable * x + self.non_trainable_variable

# Khởi tạo mô hình đơn giản
simple_module = SimpleModule(name="simple")

# Thực thi và in các kết quả kiểm tra
rs = simple_module(tf.constant(5.0))
print(rs)
print("Train variables:", simple_module.trainable_variables)
print("All variables:", simple_module.variables)

class Dense(tf.Module):
    def __init__(self, in_features, out_features, name=None):
        super().__init__(name=name)
        self.w = tf.Variable(
            tf.random.normal([in_features, out_features]), name='w')
        self.b = tf.Variable(tf.zeros([out_features]), name='b')

    def __call__(self, x):
        y = tf.matmul(x, self.w) + self.b
        return tf.nn.relu(y)

class SequentialModule(tf.Module):
    def __init__(self, name=None):
        super().__init__(name=name)

        self.dense_1 = Dense(in_features=3, out_features=3)
        self.dense_2 = Dense(in_features=3, out_features=2)

    def __call__(self, x):
        x = self.dense_1(x)
        return self.dense_2(x)

# You have made a model!
my_model = SequentialModule(name="the_model")

# Call it, with random results
print("Model results:", my_model(tf.constant([[2.0, 2.0, 2.0]])))
```

**3.19. Lưu weight**

```python
import tensorflow as tf
import numpy as np

chkp_path = "my_checkpoint"
checkpoint = tf.train.Checkpoint(model=my_model)
checkpoint.write(chkp_path)

new_model = SequentialModule()
new_checkpoint = tf.train.Checkpoint(model=new_model)
new_checkpoint.restore("my_checkpoint")

print(new_model(tf.constant([[2.0, 2.0, 2.0]])))
```

```python
import tensorflow as tf

def tao_tensor_ngau_nhien():
    x = tf.random.normal([3, 2, 2])
    y = tf.random.normal([3, 2, 2])
    return x, y

@tf.function
def tinh_toan_tren_graph(tensor_x, tensor_y):

    phep_cong = tf.add(tensor_x, tensor_y)
    phep_tru = tf.subtract(tensor_x, tensor_y)
    phep_nhan = tf.multiply(tensor_x, tensor_y)
    phep_chia = tf.divide(tensor_x, tensor_y)

    return phep_cong, phep_tru, phep_nhan, phep_chia

X, Y = tao_tensor_ngau_nhien()

ket_qua_cong, ket_qua_tru, ket_qua_nhan, ket_qua_chia = tinh_toan_tren_graph(X, Y)

print("=== TENSOR X ===")
print(X.numpy())

print("\n=== TENSOR Y ===")
print(Y.numpy())

print("\n=== KẾT QUẢ PHÉP CỘNG (GRAPH) ===")
print(ket_qua_cong.numpy())

print("\n=== KẾT QUẢ PHÉP TRỪ (GRAPH) ===")
print(ket_qua_tru.numpy())

print("\n=== KẾT QUẢ PHÉP NHÂN (GRAPH) ===")
print(ket_qua_nhan.numpy())

print("\n=== KẾT QUẢ PHÉP CHIA (GRAPH) ===")
print(ket_qua_chia.numpy())
```

**BÀI TẬP 16**

```python
import tensorflow as tf
import numpy as np

class BoxArea(tf.Module):
    @tf.function
    def __call__(self, a, b, c):
        s_xq = tf.multiply(tf.multiply(tf.add(a, b), c), 2)
        s_tp = tf.add(tf.multiply(tf.multiply(a, b), 2), s_xq)
        return s_xq, s_tp

model16 = BoxArea()
s_xung_quanh, s_toan_phan = model16(tf.constant(2.0), tf.constant(3.0), tf.constant(4.0))
print(f"Diện tích xung quanh = {s_xung_quanh}, Diện tích toàn phần = {s_toan_phan}")
```

**BÀI TẬP 17**

```python
import tensorflow as tf
import numpy as np

class TrapezoidProofSides(tf.Module):
    @tf.function
    def __call__(self, AD, BC, AB, CD):
        cond1 = tf.reduce_all(tf.equal(AD, BC))
        cond2 = tf.reduce_all(tf.equal(AB, CD))
        return tf.logical_and(cond1, cond2)

model17 = TrapezoidProofSides()
print("Kết quả chứng minh:", model17(tf.constant(5), tf.constant(5), tf.constant(4), tf.constant(4)))
```

**BÀI TẬP 18**

```python
import tensorflow as tf
import numpy as np

class TrapezoidProofBases(tf.Module):
    @tf.function
    def __call__(self, AD, BC, is_parallel_AD_BC):
        cond1 = tf.reduce_all(tf.equal(AD, BC))
        return tf.logical_and(cond1, is_parallel_AD_BC)

model18 = TrapezoidProofBases()
print("Kết quả chứng minh:", model18(tf.constant(6), tf.constant(6), tf.constant(True)))
```

**"BÀI TẬP 19**

```python
import tensorflow as tf
import numpy as np

class TriangleMidline(tf.Module):
    @tf.function
    def __call__(self, AM, MB, is_parallel_MN_BC):

        is_M_midpoint = tf.reduce_all(tf.equal(AM, MB))
        is_N_midpoint = tf.logical_and(is_M_midpoint, is_parallel_MN_BC)
        return is_N_midpoint

model19 = TriangleMidline()
print("Chứng minh NA = NC thành công:", model19(tf.constant(3.5), tf.constant(3.5), tf.constant(True)))
```
