**3.11. Tạo tensor với giá trị ngẫu nhiên**

```python
import tensorflow as tf
import numpy as np

# Hàm random.normal
range_tensor_n1 = tf.random.normal([5])
print(range_tensor_n1)
print("\n")
range_tensor_n2 = tf.random.normal([3, 3])
print(range_tensor_n2)
print("\n")

# Hàm random.truncated_normal
range_tensor_t1 = tf.random.truncated_normal([5])
print(range_tensor_t1)
print("\n")
range_tensor_t2 = tf.random.truncated_normal([3, 2])
print(range_tensor_t2)
print("\n")
range_tensor_t3 = tf.random.truncated_normal([3, 2, 2])
print(range_tensor_t3)
print("\n")

# Hàm random.uniform
range_tensor_u1 = tf.random.uniform([5], minval=10, maxval=20)
print(range_tensor_u1)
print("\n")
range_tensor_u2 = tf.random.uniform([3, 2], minval=10, maxval=20)
print(range_tensor_u2)
print("\n")
range_tensor_u3 = tf.random.uniform([3, 2, 5], minval=10, maxval=20)
print(range_tensor_u3)
print("\n")

# Ngẫu nhiên kết hợp Shuffle xáo trộn
range_tensor_shuffle = tf.random.uniform([5], minval=10, maxval=20)
print("Gốc trước khi shuffle:", range_tensor_shuffle)
print("Sau khi shuffle:", tf.random.shuffle(range_tensor_shuffle))
```

**3.12. Chuyển đổi kiểu dữ liệu**

```python
import tensorflow as tf
import numpy as np

t_float = tf.constant(3.9, tf.float32)
print(t_float, '---', t_float.dtype)
print("\n")

t_int = tf.cast(t_float, dtype=tf.int32)
print(t_int, '---', t_int.dtype)
print("\n")
```

**3.13. Các phép toán trong tensorflow**

```python
import tensorflow as tf
import numpy as np

x = tf.constant(2, dtype=tf.float32)
print("Căn bậc 2 của 2:", tf.sqrt(x), end="\n\n")

a = tf.constant([[1, 2]], tf.int32)
b = tf.constant([[3, 4]], tf.int32)
print("Cộng dùng tf.add:", tf.add(a, b), end="\n\n")
print("Cộng dùng toán tử +:", a + b, end="\n\n")

a_mul = tf.constant([[1, 2]], tf.int32)
b_mul = tf.constant([[3, 4]], tf.int32)
print("Nhân dùng tf.multiply:", tf.multiply(a_mul, b_mul), end="\n\n")
print("Nhân dùng toán tử *:", a_mul * b_mul, end="\n\n")

# Các phép toán mở rộng trên Vector
v1 = tf.constant([6., 6., 6.])
v2 = tf.constant([4., 4., 4.])
v3 = tf.constant([3., 3., 3.])

print("tf.scalar_mul(3, v1):", tf.scalar_mul(3, v1), end="\n\n")
print("tf.negative(v1):", tf.negative(v1), end="\n\n")
print("tf.add_n([v1, v2, v3]):", tf.add_n([v1, v2, v3]), end="\n\n")
```
